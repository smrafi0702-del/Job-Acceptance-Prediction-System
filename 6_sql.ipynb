```python

import pandas as pd
from sqlalchemy import create_engine

# ---- 1. Update PostgreSQL details ----
USERNAME = "postgres"
PASSWORD = "2014"
HOST = "localhost"
PORT = 5432
DATABASE = "job_placement_db"
TABLE_NAME = "job_placement_cleaned"
SCHEMA = "public"  # default PostgreSQL schema

# ---- 2. Load the cleaned CSV produced by the notebook ----
df = pd.read_csv("cleaned_job_placement_data.csv")
print(f"Loaded cleaned dataset: {df.shape[0]} rows, {df.shape[1]} columns")

# ---- 3. Connect to PostgreSQL and upload the table ----
connection_string = f"postgresql+psycopg2://{USERNAME}:{PASSWORD}@{HOST}:{PORT}/{DATABASE}"
engine = create_engine(connection_string)

df.to_sql(TABLE_NAME, con=engine, schema=SCHEMA, if_exists="replace", index=False)

print(f"✅ Data successfully uploaded to PostgreSQL table '{SCHEMA}.{TABLE_NAME}' in database '{DATABASE}'")

# ---- 4. (Optional) Quick sanity check: read the row count back ----
with engine.connect() as conn:
    result = conn.exec_driver_sql(f"SELECT COUNT(*) FROM {SCHEMA}.{TABLE_NAME};")
    count = result.scalar()
    print(f"Row count in PostgreSQL table: {count}")
```
