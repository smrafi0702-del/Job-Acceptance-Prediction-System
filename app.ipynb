## Handling Missing Values

```python
# Check missing values
missing = df.isnull().sum()

# Show only columns that have missing values
missing = missing[missing > 0]

# Calculate missing percentage
missing_percentage = (missing / len(df)) * 100

missing_summary = pd.DataFrame({
    "Missing Count": missing,
    "Missing Percentage": missing_percentage.round(2)
})

display(missing_summary)
```

```python
# Fill missing numerical values with the median
numeric_missing_cols = [
    "ssc_percentage",
    "hsc_percentage",
    "notice_period_days",
    "employment_gap_months"
]

for col in numeric_missing_cols:
    median_value = (df[col].median())
    df[col] = df[col].fillna(median_value)
    print(f"Filled missing values in '{col}' with median = {median_value:.2f}")
```

```python
categorical_missing_cols = [
    "career_switch_willingness",
    "relevant_experience",
    "job_role_match",
    "layoff_history",
    "relocation_willingness"
]

for col in categorical_missing_cols:
    mode_val = df[col].mode()[0]
    df[col] = df[col].fillna(mode_val)
    print(f"Filled missing values in '{col}' with mode = '{mode_val}'")
```

```python
# Flag records where work experience seems unusually high for the candidate's age
inconsistent_mask = (df['age_years'] - df['years_of_experience']) < 20

# Count the records that do not satisfy the age-experience rule
print("Rows with implausible age vs experience combination:", inconsistent_mask.sum())

# Display a few flagged records for manual inspection
df.loc[inconsistent_mask, ['age_years', 'years_of_experience']].head()
```

```python
# Calculate the maximum allowed experience using age 20 as the assumed starting age
max_plausible_experience = (df['age_years'] - 20).clip(lower=0)

# Reduce experience values that exceed the calculated limit
df['years_of_experience'] = np.minimum(
    df['years_of_experience'],
    max_plausible_experience
)

# Count records that still violate the age-experience condition
still_inconsistent = ((df['age_years'] - df['years_of_experience']) < 20).sum()

# Display the number of remaining inconsistent records
print(f"Remaining implausible rows after fix: {still_inconsistent}")
```

```python
# Select columns whose values should stay within the 0–100 range
pct_cols = ['ssc_percentage', 'hsc_percentage', 'degree_percentage', 'skills_match_percentage']

# Find and count any records that fall outside the permitted range
for col in pct_cols:
    out_of_range = df[(df[col] < 0) | (df[col] > 100)]
    print(f"{col}: {len(out_of_range)} rows out of valid 0-100 range")
```

```python
# Convert work experience from years into months for comparison
# Flag records where the employment gap exceeds total experience by more than 24 months
experience_in_months = df['years_of_experience'] * 12
inconsistent_gap = df[df['employment_gap_months'] > experience_in_months + 24]

print(f"Rows where employment gap exceeds experience + 24-month buffer: {len(inconsistent_gap)}")

df[['years_of_experience', 'employment_gap_months']].describe()
```

```python
# Select columns where negative values would be invalid
positive_only_cols = ['certifications_count', 'notice_period_days', 'previous_ctc_lpa', 'expected_ctc_lpa']

# Count and report records containing negative values in each selected column
for col in positive_only_cols:
    negative_rows = df[df[col] < 0]
    print(f"{col}: {len(negative_rows)} negative rows")
```

# Analyst Tasks (EDA & ML Analytics)

## 1.Candidate Performance Analysis

### 1.1 Academic Scores vs Placement Outcome

```python
# Create three side-by-side charts for academic percentages
fig, axes = plt.subplots(1, 3, figsize=(16, 5))

# Compare each academic percentage across placement outcomes
for ax, col in zip(axes, ['ssc_percentage', 'hsc_percentage', 'degree_percentage']):
    sns.boxplot(
        data=df,
        x='status',
        y=col,
        hue='status',
        palette=status_colors,
        dodge=False,
        legend=False,
        ax=ax
    )

    # Add a descriptive title to each chart
    ax.set_title(f"{col.replace('_', ' ').title()} vs Status")

# Adjust spacing and display the charts
plt.tight_layout()
plt.show()
```

```python
# Overall academic average vs placement, as a distribution
df['academic_avg_temp'] = (df['ssc_percentage'] + df['hsc_percentage'] + df['degree_percentage']) / 3

plt.figure(figsize=(8,5))
sns.histplot(data=df, x='academic_avg_temp', hue='status', kde=True, bins=30,
             palette='mako')
plt.xlabel("Average Academic Score (SSC + HSC + Degree) / 3")
plt.title("Academic Performance Distribution by Placement Status")
plt.show()

df.drop(columns=['academic_avg_temp'], inplace=True)
```


Observation: Academic scores (SSC, HSC, degree %) show only a mild difference between Placed and Not Placed candidates — academics matter, but nowhere near as strongly as technical/interview performance . This is a useful insight: strong academics alone don't guarantee offer acceptance.

### 1.2 Skills Match % vs Interview Performance

```python
df['interview_avg_temp'] = (df['technical_score'] + df['aptitude_score'] + df['communication_score']) / 3

plt.figure(figsize=(8,6))
sns.scatterplot(data=df.sample(3000, random_state=42), x='skills_match_percentage',
                 y='interview_avg_temp', hue='status', alpha=0.5,
                 palette='crest')
plt.xlabel("Skills Match %")
plt.ylabel("Average Interview Score (Technical+Aptitude+Communication)/3")
plt.title("Skills Match % vs Interview Performance")
plt.show()

print("Correlation between skills match % and average interview score:",
      round(df['skills_match_percentage'].corr(df['interview_avg_temp']), 3))

df.drop(columns=['interview_avg_temp'], inplace=True)
```


Observation: Skills match % and interview performance are positively correlated — candidates whose skills match the job well also tend to interview better. Placed candidates (blue) cluster toward the top-right (high skills match + high interview score)

### 1.3 Certification Impact on Job Acceptance

```python
cert_rate = df.groupby('certifications_count')['status'].apply(lambda x: (x=='Placed').mean()*100)

plt.figure(figsize=(8,5))
cert_rate.plot(marker='o', color='skyblue')
plt.xlabel("Number of Certifications")
plt.ylabel("Placement / Acceptance Rate (%)")
plt.title("Certification Count vs Job Acceptance Rate")
plt.show()
```

Observation: Acceptance rate generally rises with more certifications, though the effect is modest compared to technical score or skills match — certifications help, but they're a secondary signal, not the main driver.

## 3. Interview & Evaluation Analysis

### 3.1 Interview Score vs Placement Probability

```python
df['interview_avg_temp'] = (df['technical_score'] + df['aptitude_score'] + df['communication_score']) / 3
df['interview_score_bin'] = pd.cut(df['interview_avg_temp'], bins=[0,40,50,60,70,80,90,100])

bin_rate = df.groupby('interview_score_bin', observed=True)['status'].apply(lambda x: (x=='Placed').mean()*100)

plt.figure(figsize=(8,5))
bin_rate.plot(kind='bar', color='teal')
plt.xlabel("Average Interview Score Range")
plt.ylabel("Placement Probability (%)")
plt.title("Interview Score vs Placement Probability")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()

df.drop(columns=['interview_avg_temp', 'interview_score_bin'], inplace=True)
```

Observation: Placement probability rises sharply as average interview score increases — candidates scoring above 80 are placed at a much higher rate than those scoring below 50. This is a near-linear relationship, confirming interview performance is one of the most reliable predictors.

### 3.2 Employability Test Score Analysis

The dataset does not contain a dedicated column for an employability test score. Therefore, we calculate an overall employability score by taking the average of the technical, aptitude, and communication scores to represent the candidate’s combined assessment performance.

```python
df['employability_score'] = (df['technical_score'] + df['aptitude_score'] + df['communication_score']) / 3

plt.figure(figsize=(8,5))
sns.histplot(data=df, x='employability_score', hue='status', kde=True, bins=30,
             palette=status_colors)
plt.title("Employability Score Distribution by Placement Status")
plt.xlabel("Employability Score (avg of technical + aptitude + communication)")
plt.show()

print(df.groupby('status')['employability_score'].describe()[['mean','std','min','max']])

df.drop(columns=['employability_score'], inplace=True)
```

Observation: Placed candidates have a visibly higher and more tightly clustered employability score distribution, while Not Placed candidates skew lower and are more spread out — reinforcing that overall assessment performance (not just one single test) is a strong signal of acceptance.
